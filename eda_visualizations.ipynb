```python
import matplotlib.pyplot as plt
import numpy as np

# Data
age_groups = ['0–15', '16–30', '31–45', '46–60', '61+']
counts = [417, 1007, 1564, 1142, 1116]
total = sum(counts)

# Calculate percentages
percentages = [(c/total)*100 for c in counts]

# Create figure
plt.figure(figsize=(9,5))
bars = plt.bar(age_groups, counts)

plt.title('Age Distribution of Patients', fontsize=14, fontweight='bold')
plt.xlabel('Age Group', fontsize=12)
plt.ylabel('Number of Patients', fontsize=12)

# Add values + percentages on bars
for i, bar in enumerate(bars):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             height + 40,
             f'{counts[i]}\n({percentages[i]:.1f}%)',
             ha='center',
             fontsize=10)

# Add total inside plot
plt.text(3.5, max(counts)*0.9,
         f'Total Patients: {total}',
         fontsize=11,
         bbox=dict(facecolor='white', alpha=0.8))

plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
age_groups = ['0–15', '16–30', '31–45', '46–60', '61+']
counts = [417, 1007, 1564, 1142, 1116]
total = sum(counts)

# Function to show count + percentage
def autopct_format(values):
    def my_format(pct):
        count = int(round(pct * total / 100.0))
        return f'{pct:.1f}%\n({count})'
    return my_format

plt.figure(figsize=(7,7))

wedges, texts, autotexts = plt.pie(
    counts,
    labels=age_groups,
    autopct=autopct_format(counts),
    startangle=90
)

plt.title('Age Distribution of Patients', fontsize=14, fontweight='bold')

# Add total in center
plt.text(0, 0, f'Total\n{total}', ha='center', va='center', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data (English names only)
diagnoses = [
    "Pale Complexion",
    "Digestive System Disease",
    "Upper Respiratory Tract Infection",
    "Lower Respiratory Tract Infection",
    "Lower Back and Leg Pain",
    "Anemia",
    "Gastric & Duodenal Ulcer",
    "Chronic Conjunctivitis & Corneal Injury",
    "Elevated White Blood Cells",
    "Localized Limb Pain"
]

counts = [3, 3, 3, 3, 3, 3, 3, 2, 2, 2]

plt.figure(figsize=(10,6))
bars = plt.barh(diagnoses, counts)

plt.title("Top 10 Diagnoses", fontsize=14, fontweight='bold')
plt.xlabel("Number of Patients")
plt.ylabel("Diagnosis")

# Add numbers next to bars
for i, bar in enumerate(bars):
    width = bar.get_width()
    plt.text(width + 0.05, bar.get_y() + bar.get_height()/2,
             str(counts[i]),
             va='center')

plt.gca().invert_yaxis()  # Highest on top
plt.tight_layout()
plt.show()
```
