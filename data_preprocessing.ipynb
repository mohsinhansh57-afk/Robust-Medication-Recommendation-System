```python
import re

clean_path = r"/Users/ron.mos/Desktop/Jp/TraceDR-main/TraceDR-model/data/DrugRec0716/DrugRec_clean.json"

bins = {
    "0-15": 0,
    "16-30": 0,
    "31-45": 0,
    "46-60": 0,
    "61+": 0
}

pattern = re.compile(r'"age"\s*:\s*(\d+)')

with open(clean_path, "r", encoding="utf-8", errors="ignore") as f:
    for line in f:
        m = pattern.search(line)
        if m:
            age = int(m.group(1))

            if 0 <= age <= 15:
                bins["0-15"] += 1
            elif 16 <= age <= 30:
                bins["16-30"] += 1
            elif 31 <= age <= 45:
                bins["31-45"] += 1
            elif 46 <= age <= 60:
                bins["46-60"] += 1
            elif age >= 61:
                bins["61+"] += 1

print("Age Distribution:")
for k, v in bins.items():
    print(k, ":", v)

print("Total patients counted:", sum(bins.values()))
```

```python
import re
from collections import Counter

path = r"/Users/ron.mos/Desktop/Jp/TraceDR-main/TraceDR-model/data/DrugRec0716/DrugRec.json"

with open(path, "r", encoding="utf-8", errors="ignore") as f:
    content = f.read()

diagnosis_counter = Counter()

# نبحث عن كل diagnosis blocks
matches = re.findall(r'"diagnosis"\s*:\s*\[(.*?)\]', content, re.DOTALL)

for match in matches:
    diseases = re.findall(r'"(.*?)"', match)
    for d in diseases:
        diagnosis_counter[d] += 1

print("Top 10 Diagnoses:")
for disease, count in diagnosis_counter.most_common(10):
    print(disease, ":", count)

print("Total unique diagnoses:", len(diagnosis_counter))
```

```python
import re
from collections import Counter

path = r"/Users/ron.mos/Desktop/Jp/TraceDR-main/TraceDR-model/data/DrugRec0716/DrugRec.json"

with open(path, "r", encoding="utf-8", errors="ignore") as f:
    content = f.read()

medicine_counter = Counter()

# نبحث عن كل بلوكات medicine
matches = re.findall(r'"medicine"\s*:\s*\[(.*?)\]', content, re.DOTALL)

for match in matches:
    # نبحث عن اسم الدواء داخل كل object
    names = re.findall(r'"name"\s*:\s*"(.*?)"', match)
    for name in names:
        medicine_counter[name] += 1

print("Top 10 Medicines:")
for med, count in medicine_counter.most_common(10):
    print(med, ":", count)

print("Total unique medicines:", len(medicine_counter))
```

```python
import re

path = r"/Users/ron.mos/Desktop/Jp/TraceDR-main/TraceDR-model/data/DrugRec0716/DrugRec.json"

with open(path, "r", encoding="utf-8", errors="ignore") as f:
    content = f.read()

# نلقط محتوى conflict داخل []
conflicts = re.findall(r'"conflict"\s*:\s*\[(.*?)\]', content, re.DOTALL)

has_conflict = 0
no_conflict = 0

for c in conflicts:
    # نشيل المسافات والأسطر
    inside = re.sub(r"\s+", "", c)
    
    if inside == "":          # [] فاضي
        no_conflict += 1
    else:                     # فيه شيء داخل []
        has_conflict += 1

total = has_conflict + no_conflict

print("Conflict Analysis:")
print("Has Conflict:", has_conflict)
print("No Conflict:", no_conflict)
print("Total counted:", total)
```

```python
import re
from collections import Counter

path = r"/Users/ron.mos/Desktop/Jp/TraceDR-main/TraceDR-model/data/DrugRec0716/DrugRec.json"

with open(path, "r", encoding="utf-8", errors="ignore") as f:
    content = f.read()

# نلقط كل medicine_num
medicine_nums = re.findall(r'"medicine_num"\s*:\s*(\d+)', content)

# نحولهم لأرقام
medicine_nums = [int(num) for num in medicine_nums]

total_patients = len(medicine_nums)

print("Medicine Count Distribution:")
print("Total Patients:", total_patients)
print("Minimum medicine_num:", min(medicine_nums))
print("Maximum medicine_num:", max(medicine_nums))
print("Average medicine_num:", sum(medicine_nums)/total_patients)
```

```python
bins = {
    "0-10": 0,
    "11-50": 0,
    "51-100": 0,
    "101-500": 0,
    "500+": 0
}

for num in medicine_nums:
    if num <= 10:
        bins["0-10"] += 1
    elif num <= 50:
        bins["11-50"] += 1
    elif num <= 100:
        bins["51-100"] += 1
    elif num <= 500:
        bins["101-500"] += 1
    else:
        bins["500+"] += 1

print("\nMedicine Count Range Distribution:")
for k, v in bins.items():
    print(k, ":", v)
```

```python
import re

path = r"/Users/ron.mos/Desktop/Jp/TraceDR-main/TraceDR-model/data/DrugRec0716/DrugRec.json"

with open(path, "r", encoding="utf-8", errors="ignore") as f:
    content = f.read()

# نلقط محتوى antecedents داخل []
antecedents_list = re.findall(r'"antecedents"\s*:\s*\[(.*?)\]', content, re.DOTALL)

has_history = 0
no_history = 0

for a in antecedents_list:
    inside = re.sub(r"\s+", "", a)
    
    if inside == "":
        no_history += 1
    else:
        has_history += 1

total = has_history + no_history

print("Antecedents (Medical History) Analysis:")
print("With Medical History:", has_history)
print("Without Medical History:", no_history)
print("Total counted:", total)
```
